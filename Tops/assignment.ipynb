```python
def calculate_standard_deviation(score):
    mean = sum(score) / len(score)
    
    squared_diff = [(x-mean)**2 for x in score]
    
    variance = sum(squared_diff) / len(score)
    
    std = variance ** 0.5
    
    return round(std,2)

scores = [70,80,90,60,100]

print(calculate_standard_deviation(scores))
```

```python
def probability_both(food_users,dessert_users):
    food_users = set(food_users)
    dessert_users = set(dessert_users)
    
    intersection = food_users & dessert_users
    all_users = food_users | dessert_users
    
    probability = len(intersection) / len(all_users)
    
    return probability

food_users = [1,2,3,4,5,6]
dessert_users = [3,4,5,7,8]

result = probability_both(food_users,dessert_users)

print("Probability of ordering both:",result)
```
